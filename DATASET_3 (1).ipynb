#  Exploratory Data Analysis On Numerical Features

## Numerical Feature 1: Lead Time

```python
avg_lead_time = df['LEAD_TIME'].mean()
print("Average Lead Time:", round(avg_lead_time, 2), "days")
```

This above cell shows on an average how many days early people do their booking before arrival

```python
import matplotlib.pyplot as plt
plt.hist(df['LEAD_TIME'], bins=50)
plt.title("Distribution of Lead Time")
plt.xlabel("Lead Time (days)")
plt.ylabel("Number of Bookings")
plt.show()
```

This Graph shows Numbers of booking (Early vs Late)

```python
def categorize_lead_time(x):
    if x <= 7:
        return "Last-Minute"
    elif x <= 30:
        return "Short-Term"
    elif x <= 90:
        return "Medium-Term"
    else:
        return "Long-Term"
df['LEAD_TIME_category'] = df['LEAD_TIME'].apply(categorize_lead_time)
```

```python
print(df['LEAD_TIME_category'].value_counts())
```

```python
import seaborn as sns
sns.countplot(x='LEAD_TIME_category', data=df)
plt.title("Booking Type Distribution (Early vs Last-Minute)")
plt.xlabel("Booking Type")
plt.ylabel("Count")
plt.show()
```

This category helps to understand how many guests do thier booking at what interval of time

## Relation Between Lead Time & Cancellation 

```python
sns.barplot(x='IS_CANCELED', y='LEAD_TIME', data=df, errorbar=None)
plt.title("Lead Time vs Cancellation")
plt.xlabel("0 = Not Canceled, 1 = Canceled")
plt.ylabel("Lead Time")
plt.show()
```

From the above graph we can see that Higher lead time means more chances of cancellation

## Numerical Feature 2: TOTAL_OF_SPECIAL_REQUESTS

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
sns.countplot(x='TOTAL_OF_SPECIAL_REQUESTS', data=df)
plt.title("Number of Special Requests Distribution")
plt.xlabel("Number of Special Requests")
plt.ylabel("Count of Bookings")
plt.show()
```

This above graph shows the Number of Special Requests made by the Guests 

## Relation Between Total of Special Requests & Cancellation

```python
sns.barplot(x='TOTAL_OF_SPECIAL_REQUESTS', y='IS_CANCELED', data=df, errorbar=None)
plt.xlabel('No. of Special Requests Made by Guest')
plt.ylabel('Cancellation Rate')
plt.title('Special Requests vs Booking Cancellation')
plt.show()
```

The above graph gives the insight that more number of special request made by guests means less cancellations

# Exploratory Data Analysis On Categorical Features

## Categorical Feature 1: Hotel Type 

```python
sns.countplot(x='HOTEL', data=df)
plt.title("HOTEL BOOKING DISTRIBUTION")
plt.xlabel("Hotel Type")
plt.ylabel("No. of Bookings ")
plt.show()
```

This above Graph shows that City Hotel is much more preffered by the Guests

## Relation Between Hotel Type & Cancellation

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
sns.barplot(x='HOTEL', y='IS_CANCELED', data=df, errorbar=None)
plt.title("Cancellation Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Cancellation Rate")
plt.show()
```

The above graph shows that City Hotels have a higher cancellation rate

## Categorical Feature 2: Market Segment 

```python
sns.countplot(x='MARKET_SEGMENT', data=df)
plt.title("MARKET_SEGMENT")
plt.xticks(rotation=45)
plt.xlabel("Booking Channel")
plt.ylabel("No. of Bookings")
plt.show()
```

The above graph shows that Online TA (Travel Agents) is the dominant booking channel followed by Offline TA/TO and Groups.

## Relation Between MARKET SEGMENT & Cancellation

```python
sns.countplot(x='MARKET_SEGMENT', hue='IS_CANCELED', data=df)
plt.xlabel('MARKET SEGMENT Type')
plt.ylabel('No. of Bookings')
plt.title("Market Segment vs Cancellation")
plt.xticks(rotation=45)
plt.legend(title='Booking Status', labels=['Not Canceled', 'Canceled'])
plt.show()
```

This Above graph shows that Online TA (Travel Agents) booking channel have the highest cancellation rate followed by the Groups
